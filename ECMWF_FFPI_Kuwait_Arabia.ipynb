```python
import cfgrib
import xarray as xr

steps = range(3, 73, 3)

rain_rates = []
soil_moistures = []

tp_72 = None
ro_72 = None

for step in steps:
    filename = f"ffpi_{step:03d}h.grib2"

    dsets = cfgrib.open_datasets(
        filename,
        backend_kwargs={"indexpath": ""}
    )

    # نبحث عن المجموعات حسب اسم المتغير بدل الاعتماد على ترتيبها
    vsw_ds_i = next(ds for ds in dsets if "vsw" in ds.data_vars)
    main_ds_i = next(ds for ds in dsets if "tprate" in ds.data_vars)

    # قص نطاق الجزيرة العربية إلى الهند
    main_reg = main_ds_i.sel(
        latitude=slice(40, 5),
        longitude=slice(34, 100)
    )

    vsw_reg = vsw_ds_i["vsw"].sel(
        latitude=slice(40, 5),
        longitude=slice(34, 100)
    )

    # معدل المطر mm/h
    rate = main_reg["tprate"] * 3600
    rain_rates.append(rate)

    # رطوبة التربة السطحية:
    # 70% للطبقة 1 + 30% للطبقة 2
    soil = (
        0.70 * vsw_reg.sel(soilLayer=1)
        + 0.30 * vsw_reg.sel(soilLayer=2)
    )
    soil_moistures.append(soil)

    # tp و ro تراكميان، لذلك نأخذ ملف +72
    if step == 72:
        tp_72 = main_reg["tp"] * 1000
        ro_72 = main_reg["ro"] * 1000

# أعلى شدة مطر خلال جميع الخطوات
max_rain_rate = xr.concat(
    rain_rates,
    dim="forecast_step"
).max("forecast_step")

# أعلى رطوبة سطحية خلال الفترة
max_soil = xr.concat(
    soil_moistures,
    dim="forecast_step"
).max("forecast_step")

print("✅ تم تحليل جميع خطوات +3 إلى +72")
print("Max rain rate:", round(float(max_rain_rate.max()), 1), "mm/h")
print("Max TP:", round(float(tp_72.max()), 1), "mm")
print("Max runoff:", round(float(ro_72.max()), 1), "mm")
print("Max weighted soil moisture:", round(float(max_soil.max()), 3), "m3/m3")
```

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from scipy.ndimage import label

# ==========================================
# Unbiased Experimental FFPI – ECMWF
# Fixed physical thresholds
# ==========================================

# 1) تحويل المتغيرات إلى درجات 0–1 بعتبات ثابتة

# المطر التراكمي:
# 5 mm = بداية الإشارة
# 80 mm = درجة عالية جدًا
tp_score = ((tp_72 - 5) / (80 - 5)).clip(0, 1)

# أقصى شدة مطر:
# 2 mm/h = ضعيف
# 30 mm/h = شديد جدًا
rate_score = ((max_rain_rate - 2) / (30 - 2)).clip(0, 1)

# الجريان:
# 2 mm = ضعيف
# 50 mm = شديد
runoff_score = ((ro_72 - 2) / (50 - 2)).clip(0, 1)

# رطوبة التربة السطحية الموزونة:
# 0.10 = جافة
# 0.40 = رطبة جدًا
soil_score = ((max_soil - 0.10) / (0.40 - 0.10)).clip(0, 1)

# 2) الأوزان
ffpi = (
    0.40 * rate_score +
    0.25 * tp_score +
    0.25 * runoff_score +
    0.10 * soil_score
)

# 3) تجاهل الأماكن ذات المطر الضعيف
ffpi = ffpi.where(tp_72 >= 5)

# 4) حذف البكسلات/التجمعات الصغيرة جدًا
risk_mask = np.isfinite(ffpi.values) & (ffpi.values >= 0.20)

labels, nlabels = label(risk_mask)

clean_mask = np.zeros_like(risk_mask, dtype=bool)

for n in range(1, nlabels + 1):
    component = labels == n

    # الاحتفاظ فقط بالتجمعات من 3 بكسلات فأكثر
    if component.sum() >= 3:
        clean_mask |= component

ffpi_clean = ffpi.where(clean_mask)

# 5) تحويل المؤشر إلى 0–100
ffpi100 = ffpi_clean * 100

print("✅ تم حساب Unbiased Experimental FFPI")
print("أعلى قيمة للمؤشر:", round(float(ffpi100.max()), 1), "/ 100")

# ==========================================
# رسم الخريطة
# ==========================================

fig = plt.figure(figsize=(15, 8))

ax = plt.axes(
    projection=ccrs.PlateCarree()
)

ax.set_extent(
    [34, 100, 5, 40],
    crs=ccrs.PlateCarree()
)

ax.add_feature(cfeature.LAND, alpha=0.20)
ax.add_feature(cfeature.OCEAN, alpha=0.15)
ax.add_feature(cfeature.COASTLINE, linewidth=0.7)
ax.add_feature(cfeature.BORDERS, linewidth=0.5)

levels = [0, 20, 40, 60, 80, 100]

plot = ax.contourf(
    ffpi100.longitude,
    ffpi100.latitude,
    ffpi100,
    levels=levels,
    cmap="YlOrRd",
    extend="max",
    transform=ccrs.PlateCarree()
)

cbar = plt.colorbar(
    plot,
    ax=ax,
    orientation="horizontal",
    pad=0.06,
    shrink=0.80
)

cbar.set_label("Experimental FFPI – Flood Potential Index (0–100)")

ax.set_title(
    "Unbiased Experimental FFPI – ECMWF\n"
    "Arabian Peninsula • Iraq • Iran • Afghanistan • Pakistan • India | 0–72 h",
    fontsize=14
)

ax.gridlines(
    draw_labels=True,
    linewidth=0.3,
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    "ECMWF_FFPI_0_72h.png",
    dpi=180,
    bbox_inches="tight"
)

plt.show()

print("✅ الخريطة حُفظت باسم: ECMWF_FFPI_0_72h.png")
```

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from scipy.ndimage import label

# =====================================================
# Regional Unbiased Experimental FFPI – ECMWF
# Arabian Peninsula / Dry Region vs Monsoon Region
# =====================================================

# دالة تحويل بعتبات ثابتة غير خطية
def score_piecewise(x, xp, fp):
    return xr.apply_ufunc(
        np.interp,
        x,
        kwargs={"xp": xp, "fp": fp},
        dask="allowed"
    )

lon = tp_72.longitude

# ---------------------------
# 1) عتبات المنطقة الجافة
# الكويت + الجزيرة + العراق + معظم إيران
# ---------------------------

tp_west = score_piecewise(
    tp_72,
    [0, 5, 10, 25, 40, 60],
    [0, 0, 0.25, 0.75, 1.0, 1.0]
)

rate_west = score_piecewise(
    max_rain_rate,
    [0, 2, 5, 10, 20, 30],
    [0, 0, 0.25, 0.60, 1.0, 1.0]
)

runoff_west = score_piecewise(
    ro_72,
    [0, 1, 3, 8, 15, 30],
    [0, 0, 0.20, 0.55, 0.85, 1.0]
)

soil_west = score_piecewise(
    max_soil,
    [0.05, 0.10, 0.18, 0.28, 0.40],
    [0, 0.10, 0.35, 0.70, 1.0]
)

ffpi_west = (
    0.40 * rate_west +
    0.30 * tp_west +
    0.22 * runoff_west +
    0.08 * soil_west
)

# ---------------------------
# 2) عتبات المنطقة الموسمية
# الهند وما حولها
# ---------------------------

tp_east = score_piecewise(
    tp_72,
    [0, 10, 25, 50, 80, 120],
    [0, 0, 0.18, 0.50, 0.80, 1.0]
)

rate_east = score_piecewise(
    max_rain_rate,
    [0, 3, 8, 15, 25, 40],
    [0, 0, 0.20, 0.50, 0.80, 1.0]
)

runoff_east = score_piecewise(
    ro_72,
    [0, 2, 8, 20, 40, 70],
    [0, 0, 0.18, 0.50, 0.80, 1.0]
)

soil_east = score_piecewise(
    max_soil,
    [0.10, 0.18, 0.25, 0.35, 0.45],
    [0, 0.10, 0.30, 0.65, 1.0]
)

ffpi_east = (
    0.40 * rate_east +
    0.28 * tp_east +
    0.24 * runoff_east +
    0.08 * soil_east
)

# ---------------------------
# 3) دمج تدريجي بين الغرب والشرق
# غرب 60E = عتبات جافة
# شرق 68E = عتبات موسمية
# وبينهما انتقال تدريجي
# ---------------------------

blend = ((lon - 60) / 8).clip(0, 1)

ffpi_regional = (
    ffpi_west * (1 - blend) +
    ffpi_east * blend
)

# حد أدنى للهطول
ffpi_regional = ffpi_regional.where(tp_72 >= 5)

# تحويل إلى 0–100
ffpi100_regional = ffpi_regional * 100

# ---------------------------
# 4) تنظيف خفيف فقط
# لا نحذف إشارات 25 مم في الجزيرة
# ---------------------------

# غرب 60E: نظهر من 15/100
# شرق 68E: نظهر من 22/100
display_threshold = xr.where(
    lon <= 60,
    15,
    xr.where(lon >= 68, 22, 18)
)

risk_mask = (
    np.isfinite(ffpi100_regional.values) &
    (ffpi100_regional.values >= display_threshold.values)
)

labels, nlabels = label(risk_mask)

clean_mask = np.zeros_like(risk_mask, dtype=bool)

for n in range(1, nlabels + 1):
    component = labels == n

    # تنظيف خفيف: نحتفظ بتجمعات 3 بكسلات فأكثر
    if component.sum() >= 3:
        clean_mask |= component

ffpi_final = ffpi100_regional.where(clean_mask)

print("✅ تم حساب Regional Unbiased FFPI")
print(
    "أعلى قيمة:",
    round(float(ffpi_final.max()), 1),
    "/ 100"
)

# =====================================================
# الرسم
# =====================================================

fig = plt.figure(figsize=(15, 8))
ax = plt.axes(projection=ccrs.PlateCarree())

ax.set_extent(
    [34, 100, 5, 40],
    crs=ccrs.PlateCarree()
)

ax.add_feature(cfeature.LAND, alpha=0.20)
ax.add_feature(cfeature.OCEAN, alpha=0.15)
ax.add_feature(cfeature.COASTLINE, linewidth=0.7)
ax.add_feature(cfeature.BORDERS, linewidth=0.5)

levels = [15, 25, 40, 60, 80, 100]

plot = ax.contourf(
    ffpi_final.longitude,
    ffpi_final.latitude,
    ffpi_final,
    levels=levels,
    cmap="YlOrRd",
    extend="max",
    transform=ccrs.PlateCarree()
)

cbar = plt.colorbar(
    plot,
    ax=ax,
    orientation="horizontal",
    pad=0.06,
    shrink=0.82
)

cbar.set_label(
    "Regional Experimental FFPI – Flood Potential (0–100)"
)

ax.set_title(
    "Regional Unbiased Experimental FFPI – ECMWF\n"
    "Dry-region adjusted Arabian Peninsula • Monsoon-adjusted India | 0–72 h",
    fontsize=14
)

ax.gridlines(
    draw_labels=True,
    linewidth=0.3,
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    "ECMWF_FFPI_REGIONAL_0_72h.png",
    dpi=180,
    bbox_inches="tight"
)

plt.show()

print("✅ حفظت الخريطة باسم:")
print("ECMWF_FFPI_REGIONAL_0_72h.png")
```

```python
def dry_region_score(rain, rate, runoff):
    # المطر التراكمي
    rain_score = np.clip((rain - 5) / (60 - 5), 0, 1)

    # شدة المطر
    rate_score = np.clip(rate / 25, 0, 1)

    # الجريان السطحي
    runoff_score = np.clip(runoff / 20, 0, 1)

    # أوزان المؤشر
    score = 100 * (
        0.45 * rain_score +
        0.35 * rate_score +
        0.20 * runoff_score
    )

    return round(float(score), 1)


print("Rain | Rate | Runoff | FFPI")
print("--------------------------------")

for rain, rate, runoff in test_cases:
    score = dry_region_score(rain, rate, runoff)
    print(
        f"{rain:>3} mm | "
        f"{rate:>2} mm/h | "
        f"{runoff:>4} mm | "
        f"{score:>5}/100"
    )
```

```python
# معايرة FFPI للجزيرة العربية والبيئات الجافة
# 25 mm أصبحت عتبة مهمة

def dry_ffpi_v2(rain, rate, runoff, soil):
    # المطر التراكمي:
    # 25 مم ≈ 50% من مكوّن المطر
    rain_score = np.clip(rain / 50.0, 0, 1)

    # شدة المطر
    rate_score = np.clip(rate / 25.0, 0, 1)

    # الجريان السطحي
    runoff_score = np.clip(runoff / 20.0, 0, 1)

    # رطوبة التربة
    # 0.10 جافة تقريباً، 0.35 رطبة جداً
    soil_score = np.clip((soil - 0.10) / 0.25, 0, 1)

    score = 100 * (
        0.45 * rain_score +
        0.30 * rate_score +
        0.15 * runoff_score +
        0.10 * soil_score
    )

    return np.clip(score, 0, 100)


# اختبار السيناريوهات مع رطوبة تربة 0.15
print("Rain | Rate | Runoff | Soil | FFPI-v2")
print("---------------------------------------")

for rain, rate, runoff in test_cases:
    score = dry_ffpi_v2(rain, rate, runoff, 0.15)

    print(
        f"{rain:>3} mm | "
        f"{rate:>2} mm/h | "
        f"{runoff:>4} mm | "
        f"0.15 | "
        f"{float(score):.1f}/100"
    )
```
